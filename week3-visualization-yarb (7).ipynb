```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(nk['charges'], nk[ 'age'] ) 
plt.title("charges Vs age")
plt.xlabel("charges")
plt.ylabel("age")
```

```python
print( " this shows that insurance charges increase with age increasement" ) 
```

```python
plt.scatter(nk['charges'] ,nk['bmi'])
plt.title(" relation between value of charges and bmi")
plt.xlabel("charges")
plt.ylabel("bmi")
```

```python
print( " this shows that insurance charges are concentrated with low bmi" ) 
```

```python
plt.figure(figsize=(8,5))
plt.hist(nk['charges'], bins=20)
plt.title('Distribution of Insurance Charges')
plt.xlabel('Charges')
plt.ylabel('Frequency')
plt.show()
```

```python
print ( " this shows that the frequency is concentrated with low range (low charges) ,with some high outliers" ) 
```

```python
avg_region = nk.groupby('region')['charges'].mean()

avg_region.plot(kind='bar')
plt.title('Average Insurance Charges by Region')
plt.xlabel('Region')
plt.ylabel('Average Charges')
plt.show()
```

```python
print("   this shows that average insurance chares differe slightly according to regions" ) 
```

```python
average_smoker = nk.groupby('smoker')['charges' ].mean()
average_smoker.plot(kind ='bar' )
plt.title(" average charges and smoking" ) 
plt.xlabel(" smoking")
plt.ylabel(" average charges" ) 
plt.show()
```

```python
print( " this shows that smokers  clearly have higher average insurance charges than non-smokers" ) 
```
