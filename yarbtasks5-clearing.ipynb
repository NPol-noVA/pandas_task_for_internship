```python
rg.isnull().sum()
#this will print the count of null values in each column  
```

```python
#we fill all units because there're many null values  
print( "null values for Data_Value_Unit are: ") 
print (rg['Data_Value_Unit'].isnull().sum()) 
rg['Data_Value_Unit'] = rg['Data_Value_Unit'].fillna('NAN')
print ("null values after filling are : ") 
print (rg['Data_Value_Unit'].isnull().sum())
```

```python
#before_after
print("missing values for geolocation were before : ") 
print(rg['GeoLocation'].isnull().sum())
print("missing values for geolocation now : ")
rg = rg.dropna(subset=['GeoLocation'])
print(rg['GeoLocation'].isnull().sum() )
```

```python
print("values that are duplicated : ") 
print(rg.duplicated().sum()) 
rg= rg.drop_duplicates()
print( " values that are duplicated now" )  
print (rg.duplicated().sum()) 
```

```python
print("Before:", rg["LocationAbbr"].dtype)

rg["LocationAbbr"] = rg["LocationAbbr"].astype("category")

print("After:", rg["LocationAbbr"].dtype) 
```
